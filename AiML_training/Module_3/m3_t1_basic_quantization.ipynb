#Task 1: Tensor Quantization and Dequantization



## Objective

Implement basic affine INT8 quantization and dequantization using NumPy. Scale and zero points are given so you can focus purely on the quantization and dequantization formulas.

##Formulas

Quantization:    q = round(x / scale) + zero_point

Dequantization:  x_dequantized = (q - zero_point) * scale

Clip range:     [-128, 127]

```python
#import the req libs

import numpy as np
```

```python
#input tensor
tensor = np.array([1.0, -0.5, 3.2, -2.8, 0.0], dtype = np.float32)

#given paramenter
scale = 0.025
zero_point = 0

print("Original tensor:",tensor)
print("scale:",scale)
print("Zero point:",zero_point)
```

```python
#Quantization part

def quantize_tensor(tensor, scale, zero_point):

    # Divide by scale, round, add zero-point, clip to [-128, 127]
    # Return as np.int8

    quantized = tensor/scale
    quantized = np.round(quantized)
    quantized = quantized + zero_point
    quantized = np.clip(quantized,-128,127)

    return quantized.astype(np.int8)
```

```python
#De-quantization part

def dequantize_tensor(quantized_tensor, scale, zero_point):

    # Subtract zero point, multiply by scale
    # Return as float


    dequantized = quantized_tensor.astype(np.float32) - zero_point
    dequantized = dequantized * scale

    return dequantized
```

```python
quantized_tensor = quantize_tensor(tensor,scale,zero_point)
dequantized_tensor = dequantize_tensor(quantized_tensor,scale,zero_point)

#absolute error
abs_error = np.abs(tensor - dequantized_tensor)

#mean absolute error
mean_abs_error = np.mean(abs_error)

#maximum absolute error
maz_abs_error = np.max(abs_error)


print("Original Tensor:",tensor)
print("Quantized Tensor (INT8):",quantized_tensor)
print("Dequantized Tensor:",dequantized_tensor)
print("Absolute Error:",abs_error)
print("Mean Absolute Error (MAE):",mean_abs_error)
print("Maximum Absolute Error:",maz_abs_error)
```

## Observations

- The floating-point tensor was successfully quantized into INT8 format using affine quantization.
- Four values were reconstructed exactly after dequantization.
- The value 3.2 was clipped because its quantized value (128) exceeded the maximum INT8 limit (127).
- The clipping introduced a small reconstruction error of approximately 0.025.
- The overall Mean Absolute Error (MAE) was very small, indicating minimal information loss.
