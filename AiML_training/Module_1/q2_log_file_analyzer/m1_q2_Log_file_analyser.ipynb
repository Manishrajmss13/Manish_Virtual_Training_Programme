# Assignment 1 – Log File Analyzer

## Problem Statement

Parse application log files to identify ERROR and WARNING entries, group them by module, and produce a CSV report showing error frequency.




```python
#import the required libs

import pandas as pd
```

```python
#parsing the log file to split them to parts and convert to Dataframe
log_data=[]

with open("application.log",'r') as file:
  for line in file:
    line = line.strip()
    parts = line.split(" ",4)
    log_data.append(parts)

print("Total log entries:",len(log_data))
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#create dataframe from the log data
df = pd.DataFrame(
    log_data,
    columns=["Date","Time","Log Level","Module","Message"]
)

df.head()
```

```python
#check valueable info to see what the datatypes are for each column and see if there are any missing values
df.info()
```

```python
#we see that there are no missing values here and verifiy it using this command
df.isnull().sum()
```

```python
#Filter ERROR and WARNIGN entries

filtered_log = df[df["Log Level"].isin(["ERROR","WARNING"])]
print(filtered_log.shape) #to see the number of rows in both level
filtered_log.head()
```

```python
#filtering the count based on modules for ERROR and WARNING

filtered_log.groupby("Module").size()
```

```python
error_report = (
    filtered_log
        .groupby("Module")
        .size()
        .reset_index(name="Error/Warning Count")
        .sort_values(by="Error/Warning Count", ascending=False)
)

error_report
```

```python
total_issues = error_report["Error/Warning Count"].sum()

error_report["Percentage"] = (
    error_report["Error/Warning Count"] / total_issues * 100
).round(2)

error_report
```

```python
#save the report as csv

error_report.to_csv('error_report.csv',index=False)
print("CSV report generated successfully!!")
```

## Observations

- A total of 200 application log entries were generated and analyzed.
- The log file contained four log levels: INFO, WARNING, ERROR, and DEBUG.
- Only ERROR and WARNING log entries were selected for analysis.
- The filtered log entries were grouped by module to determine the frequency of issues.
- The generated CSV report summarizes the number of ERROR and WARNING entries for each application module.
- This analysis helps identify modules that may require further debugging and performance improvements.
